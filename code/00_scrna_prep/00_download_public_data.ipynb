# Download public scRNA-seq atlases

Downloads the CELLxGENE datasets used for the dot plots (Figure 3C, Figure S13) into `input_dir/scrna/`.

| Dataset | CELLxGENE dataset ID | Size |
|---|---|---|
| Tabula Sapiens, all cells (Tabula Sapiens Consortium, *Science* 2022) | `10df7690-6d10-4029-a47e-0f071bb2df83` | 45 GB |
| LBCL atlas (Li et al., *Cancer Cell* 2025) | `777cf301-2f1a-4429-9bb9-19173ba586d2` | 6.5 GB |
| Bone marrow, B-NHL after CAR-T (Strati et al., *Cell Rep Med* 2023) | `4f515692-9f9c-47db-889d-efc6cffc303d` | 0.7 GB |

```python
import sys; sys.path.insert(0, "../../utils")
from config import INPUT_DIR

import shutil
import urllib.request

URL = "https://datasets.cellxgene.cziscience.com/{}.h5ad"

# local file: (dataset ID, size in bytes)
DATASETS = {
    "tabula_sapiens_all_cells_10df7690.h5ad": ("10df7690-6d10-4029-a47e-0f071bb2df83", 45_144_805_889),
    "li2025_lbcl_atlas_777cf301.h5ad":        ("777cf301-2f1a-4429-9bb9-19173ba586d2", 6_505_896_722),
    "strati2023_bmmc_4f515692.h5ad":          ("4f515692-9f9c-47db-889d-efc6cffc303d", 682_628_818),
}
```

```python
outdir = INPUT_DIR / "scrna"
outdir.mkdir(parents=True, exist_ok=True)

for fname, (dataset_id, size) in DATASETS.items():
    dest = outdir / fname
    if dest.exists() and dest.stat().st_size == size:
        print(f"{fname}: already downloaded")
        continue
    print(f"{fname}: downloading {size / 1e9:.1f} GB ...", flush=True)
    with urllib.request.urlopen(URL.format(dataset_id)) as r, open(dest, "wb") as fh:
        shutil.copyfileobj(r, fh, length=1 << 23)
    assert dest.stat().st_size == size, f"{fname}: incomplete download, run again"
```
